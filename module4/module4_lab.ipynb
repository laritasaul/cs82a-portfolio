```python
import pandas as pd
df = pd.read_csv("sales_clean.csv")
df.shape
```

```python
print(df["price"].mean())
print(df["price"].median())
print(df["price"].std())
print(df["qty"].mean())
print(df["qty"].median())
print(df["qty"].std())
```

```python
#I would report the median for price. The difference between them is noticeable.
#The standard deviantion is also bigger than the mean which indicates that it is being skewed.
```

```python
df["price"].plot(kind="hist", bins=20)
```

```python
df["revenue"] = df["price"] + df["qty"]
```
