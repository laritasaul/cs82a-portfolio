```python
temps = [10, 16, 23, 28, 14, 22, 30]

for temp in temps:
#15 is the cutoff for cold temperature
    if temp < 15:
        print(temp , "is freezing")
#25 is the cutoff for cold temperature
    elif temp <= 25:
        print(temp , "is cold")
    elif temp < 27:
        print(temp , "is warm")
    else:
        temp > 27
        print(temp , "is hot")
    
```
