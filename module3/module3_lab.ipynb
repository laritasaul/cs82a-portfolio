```python
import pandas as pd
df = pd.read_csv("messy_sales.csv")
df.shape
df.head(10)
df.dtypes
df.describe
```

```python
#I see that it shows NaN for prices. Also the dates are shown in a different way for some of the columns
#There is also a negative for quantity which does not make sense.
```

```python
df.isna().sum()
```

```python
fill_value = df["price"].median()
df["price"] = df["price"].fillna(fill_value)
```

```python
df.isna().sum()
```

```python
df.shape
```

```python
print(df.duplicated().sum())
df = df.drop_duplicates()
```

```python
df.shape
```

```python
df["zip"] = df["zip"].astype(str).str.zfill(5)
```

```python
df["date"] = pd.to_datetime(df["date"], format="mixed")
```

```python
df.shape
df.head(10)
df.dtypes
df.describe
```
