```python
import pandas as pd

df = pd.read_csv("custdata.tsv", sep="\t")
print(df.columns.tolist())
print(df[["age", "income", "num.vehicles"]].describe())
```

```python

cols = ["age", "income", "num.vehicles"]
df.info()
#Before cleaning
print(df[cols].corr().round(3))
```

```python
#I took out incomes below 0, as well as ages that are 0 and over 100.
#As well as values that were NA.
clean = df[cols].dropna()
clean = clean[(clean["age"] > 0) & (clean["age"] <= 100)]
clean = clean[clean["income"] >= 0]
```

```python
#After cleaning
print(clean.corr().round(3))
clean.info()
```
