```python
import pandas as pd
import matplotlib.pyplot as plt
 
df = pd.read_csv("custdata.tsv", sep="\t")
clean = df.dropna(subset=["housing.type"])
 
counts = clean["housing.type"].value_counts()
 
fig, ax = plt.subplots(figsize=(9, 5))
bars = ax.bar(counts.index, counts.values, color="#3b6ea5")
ax.bar_label(bars, padding=3)
ax.set_title("Most customers have a mortgage or rent.")
ax.set_xlabel("Housing type")
ax.set_ylabel("Number of customers")
```
