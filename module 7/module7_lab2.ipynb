```python
import pandas as pd
from sklearn.linear_model import LinearRegression
df = pd.read_csv("ratings.tsv", sep="\t")
df.head()
```

```python
features = ["ambience", "food", "service"]
x = df[features]
y = df["rating"]
model = LinearRegression()
model.fit(x,y)
print("Intercept:", round(model.intercept_, 3))
print("Coefficients:", model.coef_.round(3))
print("R2:", round(model.score(x,y), 3))
```

```python
coefs = (pd.DataFrame({"feature": features, "coefficient":model.coef_})
.sort_values("coefficient", ascending=False)
.reset_index(drop=True))
coefs["coefficient"] = coefs["coefficient"].round(3)
coefs.index += 1
coefs
```
